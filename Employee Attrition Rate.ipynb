```python
model_df = df.copy()

# Encode target variable
model_df["Attrition"] = model_df["Attrition"].map({
    "No": 0,
    "Yes": 1
})

print(model_df.head())
```

```python
print(model_df["Attrition"].value_counts())

print("\nUnique target values:")
print(model_df["Attrition"].unique())

print("\nAttrition percentage:")
print(
    model_df["Attrition"]
    .value_counts(normalize=True) * 100
)
```

```python
columns_to_drop = [
    "EmployeeNumber",
    "EmployeeCount",
    "Over18",
    "StandardHours"
]

model_df = model_df.drop(
    columns=columns_to_drop
)

print("Updated Dataset Shape:", model_df.shape)
```

```python
X = model_df.drop(
    columns=["Attrition"]
)

y = model_df["Attrition"]

print("Independent Features:", X.shape)
print("Dependent Target:", y.shape)
```

```python
numerical_features = X.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

categorical_features = X.select_dtypes(
    include=["object"]
).columns.tolist()

print("Numerical Features:", len(numerical_features))
print("Categorical Features:", len(categorical_features))

print("\nNumerical Columns:")
print(numerical_features)

print("\nCategorical Columns:")
print(categorical_features)
```

```python
class_distribution = pd.DataFrame({
    "Employee Count": y.value_counts(),
    "Percentage": y.value_counts(normalize=True) * 100
})

class_distribution.index = [
    "Stayed" if i == 0 else "Left"
    for i in class_distribution.index
]

display(class_distribution)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=10,
    stratify=y
)

print("Training Data:", X_train.shape)
print("Testing Data:", X_test.shape)

print("\nTraining Target Distribution:")
print(y_train.value_counts())

print("\nTesting Target Distribution:")
print(y_test.value_counts())
```

```python
numerical_transformer = StandardScaler()

categorical_transformer = OneHotEncoder(
    handle_unknown="ignore"
)
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        (
            "numerical",
            numerical_transformer,
            numerical_features
        ),
        (
            "categorical",
            categorical_transformer,
            categorical_features
        )
    ]
)

print("Preprocessing pipeline created successfully!")
```

```python
logistic_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        (
            "classifier",
            LogisticRegression(
                max_iter=1000,
                random_state=10
            )
        )
    ]
)

print(logistic_model)
```

```python
logistic_model.fit(
    X_train,
    y_train
)

print("Logistic Regression Model Trained Successfully!")
```

```python
y_pred = logistic_model.predict(X_test)

y_probability = logistic_model.predict_proba(
    X_test
)[:, 1]

print("Predictions generated successfully!")

print("First 10 Predictions:")
print(y_pred[:10])
```

```python
accuracy = accuracy_score(y_test, y_pred)

precision = precision_score(
    y_test,
    y_pred,
    zero_division=0
)

recall = recall_score(
    y_test,
    y_pred,
    zero_division=0
)

f1 = f1_score(
    y_test,
    y_pred,
    zero_division=0
)

roc_auc = roc_auc_score(
    y_test,
    y_probability
)

print("LOGISTIC REGRESSION PERFORMANCE")
print("=" * 40)

print(f"Accuracy:  {accuracy * 100:.2f}%")
print(f"Precision: {precision * 100:.2f}%")
print(f"Recall:    {recall * 100:.2f}%")
print(f"F1 Score:  {f1 * 100:.2f}%")
print(f"ROC-AUC:   {roc_auc:.4f}")
```

```python
print(
    classification_report(
        y_test,
        y_pred,
        target_names=["Stayed", "Left"],
        digits=4
    )
)
```

```python
cm = confusion_matrix(
    y_test,
    y_pred,
    labels=[0, 1]
)

print(cm)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=["Stayed", "Left"]
)

disp.plot(
    cmap="Blues",
    values_format="d"
)

plt.title("Logistic Regression - Confusion Matrix")
plt.show()
```

```python
fpr, tpr, thresholds = roc_curve(
    y_test,
    y_probability
)

plt.figure(figsize=(8, 6))

plt.plot(
    fpr,
    tpr,
    label=f"Logistic Regression (AUC = {roc_auc:.3f})"
)

plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--",
    label="Random Classifier"
)

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve - Employee Attrition Prediction")

plt.legend()
plt.grid(alpha=0.3)

plt.show()
```

```python
comparison_df = pd.DataFrame({
    "Actual": y_test.values,
    "Predicted": y_pred,
    "Attrition Probability": y_probability
})

comparison_df["Actual"] = comparison_df["Actual"].map({
    0: "Stayed",
    1: "Left"
})

comparison_df["Predicted"] = comparison_df["Predicted"].map({
    0: "Stayed",
    1: "Left"
})

comparison_df.head(15)
```

```python
balanced_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        (
            "classifier",
            LogisticRegression(
                max_iter=1000,
                class_weight="balanced",
                random_state=10
            )
        )
    ]
)

balanced_model.fit(
    X_train,
    y_train
)

print("Balanced Logistic Regression Trained Successfully!")
```

```python
balanced_pred = balanced_model.predict(X_test)

balanced_probability = balanced_model.predict_proba(
    X_test
)[:, 1]

print("Balanced Predictions Generated!")
```

```python
print("BALANCED LOGISTIC REGRESSION")
print("=" * 40)

print(
    classification_report(
        y_test,
        balanced_pred,
        target_names=["Stayed", "Left"],
        digits=4
    )
)

print("Accuracy:",
      accuracy_score(y_test, balanced_pred))

print("Precision:",
      precision_score(y_test, balanced_pred))

print("Recall:",
      recall_score(y_test, balanced_pred))

print("F1 Score:",
      f1_score(y_test, balanced_pred))

print("ROC-AUC:",
      roc_auc_score(y_test, balanced_probability))
```

```python
model_comparison = pd.DataFrame({
    "Metric": [
        "Accuracy",
        "Precision",
        "Recall",
        "F1 Score",
        "ROC-AUC"
    ],

    "Logistic Regression": [
        accuracy_score(y_test, y_pred),
        precision_score(y_test, y_pred),
        recall_score(y_test, y_pred),
        f1_score(y_test, y_pred),
        roc_auc_score(y_test, y_probability)
    ],

    "Balanced Logistic Regression": [
        accuracy_score(y_test, balanced_pred),
        precision_score(y_test, balanced_pred),
        recall_score(y_test, balanced_pred),
        f1_score(y_test, balanced_pred),
        roc_auc_score(y_test, balanced_probability)
    ]
})

display(
    model_comparison.round(4)
)
```

```python
comparison_plot = model_comparison.set_index(
    "Metric"
)

comparison_plot.plot(
    kind="bar",
    figsize=(11, 6)
)

plt.title("Logistic Regression Model Comparison")
plt.ylabel("Score")
plt.xlabel("Evaluation Metrics")
plt.xticks(rotation=0)
plt.ylim(0, 1.05)
plt.legend(title="Models")

plt.tight_layout()
plt.show()
```

```python
cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=10
)

original_cv_scores = cross_val_score(
    logistic_model,
    X_train,
    y_train,
    cv=cv,
    scoring="f1"
)

balanced_cv_scores = cross_val_score(
    balanced_model,
    X_train,
    y_train,
    cv=cv,
    scoring="f1"
)

print("Original Model CV Scores:")
print(original_cv_scores)

print("\nBalanced Model CV Scores:")
print(balanced_cv_scores)

print("\nOriginal Mean F1:",
      original_cv_scores.mean())

print("Balanced Mean F1:",
      balanced_cv_scores.mean())
```

```python
cv_results = pd.DataFrame({
    "Fold": [1, 2, 3, 4, 5],
    "Logistic Regression": original_cv_scores,
    "Balanced Logistic Regression": balanced_cv_scores
})

cv_results.set_index("Fold").plot(
    kind="bar",
    figsize=(10, 5)
)

plt.title("5-Fold Cross-Validation F1-Score")
plt.ylabel("F1 Score")
plt.xlabel("Fold")
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```

```python
fitted_preprocessor = logistic_model.named_steps[
    "preprocessor"
]

feature_names = fitted_preprocessor.get_feature_names_out()

coefficients = (
    logistic_model.named_steps["classifier"].coef_[0]
)

feature_importance = pd.DataFrame({
    "Feature": feature_names,
    "Coefficient": coefficients
})

feature_importance["Absolute Coefficient"] = (
    feature_importance["Coefficient"].abs()
)

feature_importance = feature_importance.sort_values(
    by="Absolute Coefficient",
    ascending=False
)

feature_importance.head(15)
```

```python
top_features = feature_importance.head(15).sort_values(
    by="Coefficient"
)

plt.figure(figsize=(11, 7))

sns.barplot(
    data=top_features,
    x="Coefficient",
    y="Feature",
    hue="Coefficient",
    palette="coolwarm",
    legend=False
)

plt.axvline(
    0,
    color="black",
    linewidth=1
)

plt.title("Top 15 Logistic Regression Coefficients")
plt.xlabel("Coefficient")
plt.ylabel("Feature")

plt.tight_layout()
plt.show()
```

```python
if balanced_cv_scores.mean() > original_cv_scores.mean():

    final_model = balanced_model
    final_model_name = "Balanced Logistic Regression"

else:

    final_model = logistic_model
    final_model_name = "Logistic Regression"

print("Selected Model:", final_model_name)
```

```python
final_predictions = final_model.predict(X_test)

final_probabilities = final_model.predict_proba(
    X_test
)[:, 1]

final_accuracy = accuracy_score(
    y_test,
    final_predictions
)

final_precision = precision_score(
    y_test,
    final_predictions
)

final_recall = recall_score(
    y_test,
    final_predictions
)

final_f1 = f1_score(
    y_test,
    final_predictions
)

final_roc_auc = roc_auc_score(
    y_test,
    final_probabilities
)

final_results = pd.DataFrame({
    "Metric": [
        "Accuracy",
        "Precision",
        "Recall",
        "F1 Score",
        "ROC-AUC"
    ],

    "Score": [
        final_accuracy,
        final_precision,
        final_recall,
        final_f1,
        final_roc_auc
    ]
})

display(final_results.round(4))
```

```python
joblib.dump(
    final_model,
    "employee_attrition_model.pkl"
)

print("Final model saved successfully!")
```
